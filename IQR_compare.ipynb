```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset
Group_A = [45, 48, 50, 52, 49, 51, 47, 50]
Group_B = [60, 62, 65, 63, 61, 64, 66, 95]

# Function to calculate statistics
def calculate_statistics(data, name):

    mean = np.mean(data)
    median = np.median(data)
    minimum = np.min(data)
    maximum = np.max(data)

    # Sample standard deviation
    std = np.std(data, ddof=1)

    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1

    # Outlier limits
    lower_fence = Q1 - 1.5 * IQR
    upper_fence = Q3 + 1.5 * IQR

    # Find outliers
    outliers = [x for x in data
                if x < lower_fence or x > upper_fence]

    print("\n", name)
    print("-" * 30)
    print("Mean       :", mean)
    print("Median     :", median)
    print("Minimum    :", minimum)
    print("Maximum    :", maximum)
    print("Std Dev    :", round(std, 2))
    print("Q1         :", Q1)
    print("Q3         :", Q3)
    print("IQR        :", IQR)
    print("Lower Fence:", lower_fence)
    print("Upper Fence:", upper_fence)
    print("Outlier(s) :", outliers)

    return mean, median, std, outliers


# Calculate statistics
A = calculate_statistics(Group_A, "Group A")
B = calculate_statistics(Group_B, "Group B")


# Compare means
print("\n========== COMPARISON ==========")

if A[0] > B[0]:
    print("Group A has the higher mean.")
else:
    print("Group B has the higher mean.")


# Compare variation
if A[2] > B[2]:
    print("Group A has greater variation.")
else:
    print("Group B has greater variation.")


# Outlier
print("Outlier in Group B:", B[3])


# Effect of outlier on mean
Group_B_without_outlier = [60, 62, 65, 63, 61, 64, 66]

mean_without_outlier = np.mean(Group_B_without_outlier)

print("\nMean of Group B with outlier   :", np.mean(Group_B))
print("Mean of Group B without outlier:", mean_without_outlier)


# Box Plot
plt.boxplot(
    [Group_A, Group_B],
    tick_labels=["Group A", "Group B"],
    showmeans=True
)

plt.title("Box Plot - Group A vs Group B")
plt.ylabel("Values")
plt.grid(axis="y", alpha=0.3)

plt.show()
```
